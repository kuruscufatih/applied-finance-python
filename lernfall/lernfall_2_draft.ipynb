### ⚠️ Eure wichtige Entscheidung: `auto_adjust`

Die Funktion hat einen Parameter `auto_adjust`, der entweder `True` oder `False` sein kann. Euer Vorgänger hat ihn **offen gelassen**. Ihr müsst ihn selbst festlegen.

Diese Entscheidung bestimmt, **welche Kurse** ihr bekommt. Und damit, ob Frau Kellers Frage zu den Dividenden richtig beantwortet wird.

Findet selbst heraus, was dahintersteckt!

```python
AUTO_ADJUST = True # TODO: decide: True or False? Read section 3 first!


def load_prices(ticker, start, end, auto_adjust):
    """Download daily prices for one ticker from Yahoo Finance."""
    if auto_adjust not in (True, False):
        raise ValueError("Please set AUTO_ADJUST to True or False first (see section 3).")
    end_inclusive = date.fromisoformat(end) + timedelta(days=1)   # yfinance excludes the end date
    df = yf.Ticker(ticker).history(start=start, end=end_inclusive, auto_adjust=auto_adjust)
    if df.empty:
        raise ValueError(f"No data for {ticker}. Check the ticker and the dates.")
    df.index = df.index.tz_localize(None)   # remove time zone info from the dates
    return df


allianz = load_prices(portfolio[0]["ticker"], portfolio[0]["buy_date"], end_date, AUTO_ADJUST)
allianz.head()
```

## 3 · Eine Funktion für jede Aktie

Euer Code aus Teil 4 funktioniert für eine Aktie. Damit er für **jede** Aktie funktioniert, packt ihr ihn in eine Funktion.

💡 Vorgehen:

1. Kopiert eure Rechnung aus Teil 4 in die Funktion.
2. Ersetzt `allianz` durch die Variable `prices`, die in der Funktion geladen wird.
3. Gebt die Ergebnisse als Dictionary zurück.
4. Testet die Funktion mit der Allianz: Es muss dasselbe herauskommen wie in Teil 4.

```python
allianz.head()
```

```python
def analyse_stock(name, ticker, buy_date, end_date, price_column):
    """Calculate total return and CAGR for one stock."""
    prices = load_prices(ticker, buy_date, end_date, AUTO_ADJUST)
    series = prices[price_column]

    start_price = series.iloc[0]
    end_price = series.iloc[-1]
    years = (series.index[-1] - series.index[0]).days / 365.25

    total_return = end_price / start_price - 1
    cagr = (1 + total_return) ** (1 / years) - 1

    return {
        "Stock": name,
        "Total return": total_return,
        "CAGR": cagr,
    }


# Test with Allianz: must give the same result as section 4
analyse_stock("Allianz SE", "ALV.DE", "2021-09-30", end_date, "Close")
```

## 4 · Das ganze Depot und die Rangliste

Jetzt wendet ihr die Funktion in einer Schleife auf alle Aktien im `portfolio` an (siehe Crashkurs 1.4 und 1.6).

```python
results = []

for stock in portfolio:
    # TODO: call analyse_stock(...) for this stock and append the result to `results`
    stock_calculation = analyse_stock(stock["name"], stock["ticker"], stock["buy_date"], end_date, "Close")
    results.append(stock_calculation)

ranking = pd.DataFrame(results)
ranking = ranking.sort_values("CAGR", ascending=False).reset_index(drop=True)
ranking.index = ranking.index + 1
ranking
```
