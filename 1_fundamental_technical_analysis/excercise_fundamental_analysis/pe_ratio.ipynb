# Step 3: Retrieve EPS (TTM) with BeautifulSoup

We scrape **Yahoo Finance Key Statistics** page and extract EPS (TTM).  
This function will return the EPS as a float.

```python
def get_eps_ttm(ticker: str) -> float:
    """Scrape Yahoo Finance (statistics page) to get EPS (TTM) for a given ticker."""
    url = f"https://finance.yahoo.com/quote/{ticker}/"
    headers = {"User-Agent": "Mozilla/5.0"}
    response = requests.get(url, headers=headers)
    soup = BeautifulSoup(response.text, "html.parser")
    return soup.find("li", class_="").find("span", class_="").text
```

```python
get_eps_ttm("AMZN")
```
