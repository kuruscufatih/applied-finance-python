# Step 3: Calculate Book Value per Stock

```python
mercedes_total_assets = 256_627 * 1_000_000
mercedes_total_liabilities = (106_017+12_655+5_069+10_395+15_476) * 1_000_000
mercedes_outstanding_shares = 1_000_000_000
mercedes_equity = mercedes_total_assets - mercedes_total_liabilities
mercedes_book_value_per_share = mercedes_equity / mercedes_outstanding_shares
mercedes_book_value_per_share
```

# Step 4: Retrieve Stock Values for Q2 2025

```python
ticker = 'MBG.DE'
stock_data = yf.download(ticker, start='2025-06-30', end='2025-07-01')
closing_price = stock_data['Close'].iloc[0]
closing_price
```

# Step 5: Calculate P/B Ratio for Mercedes

```python
PB_RATIO_MERCEDES = closing_price / mercedes_book_value_per_share
PB_RATIO_MERCEDES
```
