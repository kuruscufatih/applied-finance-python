## 1. The Ticker Object

The **Ticker** object is the main interface for accessing data about a specific stock.

### Key Concept: What is a Ticker Symbol?

A **ticker symbol** is a unique abbreviation used to identify publicly traded companies on a stock exchange.

Examples:
- `AAPL` = Apple Inc.
- `MSFT` = Microsoft Corporation
- `GOOGL` = Alphabet Inc. (Google)
- `TSLA` = Tesla Inc.

````python
# Create a Ticker object for Apple
apple = yf.Ticker("AAPL")

print("Ticker object created for Apple (AAPL)")
print(type(apple))
````

````python
# Get company info
info = apple.info

# Display selected information
print(f"Company Name: {info.get('longName', 'N/A')}")
print(f"Sector: {info.get('sector', 'N/A')}")
print(f"Industry: {info.get('industry', 'N/A')}")
print(f"Current Price: ${info.get('currentPrice', 'N/A')}")
````

## 2. Historical Data - The Most Important Function

The `.history()` method retrieves historical market data and returns it as a **pandas DataFrame**.

### Key Concept: DataFrame

A **DataFrame** is a 2-dimensional labeled data structure (like a table) from the pandas library. It has:
- **Rows**: indexed by dates (in our case)
- **Columns**: different data fields (Open, High, Low, Close, Volume, etc.)

### Key Concept: Time Periods

You can specify different time periods:
- `period`: Use predefined periods like `"1d"`, `"5d"`, `"1mo"`, `"3mo"`, `"6mo"`, `"1y"`, `"2y"`, `"5y"`, `"10y"`, `"ytd"`, `"max"`
- `start` and `end`: Use specific dates in format `"YYYY-MM-DD"`

````python
# Get historical data for the last year
apple = yf.Ticker("AAPL")
data_1y = apple.history(period="1y")

print("Historical data for AAPL (last year):")
print(f"Data Frame: {type(data_1y)}")
print(data_1y.head())  # Show first 5 rows
print("\nDataFrame shape:", data_1y.shape)
print("Columns:", data_1y.columns.tolist())
````

````python
# Get historical data with specific start and end dates
data_custom = apple.history(start="2022-01-01", end="2024-12-31")

print(f"Historical data for AAPL from 2022-01-01 to 2024-12-31:")
print(f"Total trading days: {len(data_custom)}")
print("\nFirst 3 rows:")
print(data_custom.head(3))
print("\nLast 3 rows:")
print(data_custom.tail(3))
````

### Understanding the Columns

- **Open**: Opening price of the trading day
- **High**: Highest price during the trading day
- **Low**: Lowest price during the trading day
- **Close**: Closing price of the trading day (most commonly used)
- **Volume**: Number of shares traded
- **Dividends**: Dividend payments (if any)
- **Stock Splits**: Stock split information (if any)

````python
# Accessing specific columns
closing_prices = data_1y['Close']

print("Closing prices (last 10 days):")
print(closing_prices.tail(10))
````

### Exercise 1: Calculate Annual Averages

**Task:** Download data for 6 stocks of your choice for the last 3 years (2022, 2023, 2024) and calculate the average closing price for each stock in each year.

**Steps:**
1. Select 6 stock tickers
2. Download historical data from 2022-01-01 to 2024-12-31
3. Calculate the average closing price for each year

**Hints:**
- Use `yf.download()` to get data for multiple stocks
- Use `.resample('Y')` or filter by year to group data
- Use `.mean()` to calculate averages

````python
# Step 1: Define tickers
automobile_tickers = ["MBG.DE", "BMW.DE", "TSLA", "BYDDY"]

# Step 2: Download data
df_automobile = yf.download(automobile_tickers, start = "2022-01-01", end = "2025-11-01")

# Step 3: Calculate annual averages using groupby
df_automobile_aggregated = df_automobile.groupby(df_automobile.index.year)['Close'].mean().round(2)

# df_automobile_aggregated.plot(kind='bar', figsize=(12, 6))
df_automobile_aggregated
````
