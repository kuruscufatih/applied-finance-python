## 6.2 Conditional Selection (Boolean Indexing)

```python
df[(df["Volume"] > 40_000_000) | (df["High"] > 250)]
```

```python
# Days where closing price was above $300
high_price_days = df[df['Close'] > 300]

print(f"Number of days with Close > $300: {len(high_price_days)}")
print(f"Percentage of total days: {len(high_price_days) / len(df) * 100:.2f}%")
print("\nFirst few high-price days:")
high_price_days.head()
```

```python
# Days with high volume (above 30 million)
high_volume_days = df[df['Volume'] > 30_000_000]

print(f"Number of high-volume days: {len(high_volume_days)}")
high_volume_days.head()
```

```python
# Multiple conditions with & (AND)
# Days where Close > $300 AND Volume > 30 million
condition = (df['Close'] > 340) & (df['Volume'] > 30_000_000)
filtered_df = df[condition]

print(f"Days with Close > $300 AND Volume > 30M: {len(filtered_df)}")
filtered_df.head()
```

```python
# Multiple conditions with | (OR)
# Days where Close > $350 OR Volume > 50 million
condition = (df['Close'] > 350) | (df['Volume'] > 50_000_000)
filtered_df = df[condition]

print(f"Days with Close > $350 OR Volume > 50M: {len(filtered_df)}")
filtered_df.head()
```

# 9. Grouping and Aggregation - Group By Year

```python
df.head()
```

```python
df_transform[["Date", "Year", "Quarter"]].head()
```

```python
df_transform.head()
```

```python
df_transform[["Year","Quarter", "Close"]].groupby(["Year", "Quarter"]).agg(
    avg_close=('Close','mean'),
    max_close=('Close', 'max')
)
```

```python
df_transform['Date'] = pd.to_datetime(df_transform['Date'], utc=True)
df_transform['Year'] = df_transform['Date'].dt.year
df_transform['Quarter'] = df_transform['Date'].dt.quarter
```

```python
df_transform[['Year','Quarter', 'Close']].groupby(['Year','Quarter']).agg(
    max_close=('Close', 'max'),
    avg_close=('Close', 'mean')
)
```

```python
yearly_stats = df_transform.groupby('Year')['Close'].agg([
    ('Mean', 'mean'),
    ('Median', 'median'),
    ('Min', 'min'),
    ('Max', 'max'),
    ('Std', 'std')
]).round(2)

print("Yearly statistics for closing prices:")
yearly_stats
```
