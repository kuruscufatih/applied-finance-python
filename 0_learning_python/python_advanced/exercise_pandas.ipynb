# Exercise 4: Grouping and Aggregation 

## Task 4.1: Date Feature Extraction

Extract the following from the Date column:
1. `Year`
2. `Month`
3. `Quarter`
4. `Day_of_Week` (0=Monday, 6=Sunday)

Display the first 15 rows with Date and all extracted features.

```python
df_work['Year'] = df_work['Date'].dt.year
df_work['Month'] = df_work['Date'].dt.month
df_work['Quarter'] = df_work['Date'].dt.quarter
df_work['Day_of_Week'] = df_work['Date'].dt.dayofweek
```

## Task 4.2: Annual Analysis 

Group the data by Year and calculate:

1. Mean, Median, Min, and Max of the Close price for each year
2. Mean daily return for each year (in %)

Present your results in a well-formatted table.

## Task 4.3: Quarterly Performance

For each quarter in each year:
1. Calculate the mean closing price
2. Calculate the total return for that quarter
3. Identify which quarter had the best performance
4. Identify which quarter had the worst performance

Create a pivot table or grouped table showing Year x Quarter.

```python
# SOLUTION
quarterly_stats = df_work.groupby(['Year', 'Quarter']).agg({
    'Close': 'mean',
    'Returns': lambda x: ((1 + x).prod() - 1) * 100
}).round(2)

quarterly_stats.columns = ['Mean_Close', 'Quarterly_Return_%']

print("Quarterly Statistics:")
display(quarterly_stats)

# Best and worst quarters
best_quarter = quarterly_stats['Quarterly_Return_%'].idxmax()
worst_quarter = quarterly_stats['Quarterly_Return_%'].idxmin()

print(f"\nBest quarter: {best_quarter} with {quarterly_stats.loc[best_quarter, 'Quarterly_Return_%']:.2f}% return")
print(f"Worst quarter: {worst_quarter} with {quarterly_stats.loc[worst_quarter, 'Quarterly_Return_%']:.2f}% return")
```

## Task 4.4: Day of Week Analysis 

Analyze returns by day of the week:
1. Calculate mean return for each day of the week
2. Calculate the percentage of positive return days for each weekday
3. Determine which day of the week had the highest average return
4. Determine which day had the most volatile returns (highest std)

Present the results in a clear table.

```python
# SOLUTION
# Create day names
day_names = {0: 'Monday', 1: 'Tuesday', 2: 'Wednesday', 3: 'Thursday', 4: 'Friday'}
df_work['Day_Name'] = df_work['Day_of_Week'].map(day_names)

# Calculate statistics by day
day_stats = df_work.groupby('Day_Name')['Returns'].agg([
    ('Mean_Return_%', lambda x: x.mean() * 100),
    ('Std_Return_%', lambda x: x.std() * 100),
    ('Positive_Days_%', lambda x: (x > 0).sum() / len(x) * 100),
    ('Count', 'count')
]).round(3)

# Reorder by day of week
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday']
day_stats = day_stats.reindex(day_order)

print("Day of Week Analysis:")
display(day_stats)

# Best and worst days
best_day = day_stats['Mean_Return_%'].idxmax()
most_volatile_day = day_stats['Std_Return_%'].idxmax()

print(f"\nBest day: {best_day} with average return of {day_stats.loc[best_day, 'Mean_Return_%']:.3f}%")
print(f"Most volatile day: {most_volatile_day} with std of {day_stats.loc[most_volatile_day, 'Std_Return_%']:.3f}%")
```
