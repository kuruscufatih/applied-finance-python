### Basic Functions

````python
monthly_closings_aapl_2020 = np.array([100,102,103,101,96,111,113,116,120,150,140,134])
monthly_closings_aapl_2021 = np.array([136,144,138,150,180,142,145,150,165,177,165,155])
````

````python
monthly_closings_aapl_2021.shape
````

````python
np.sort(monthly_closings_aapl_2020)
````

````python
np.concatenate(monthly_closings_aapl_2020, monthly_closings_aapl_2021)
````

Working with two dimensions

With .reshape() we can change the shape of an array.
Note: New Array needs to have same number of elements as original array.


````python
quarterly_closing_aapl_2020 = monthly_closings_aapl_2020.reshape(4,3) #quarterly 
quarterly_closing_aapl_2020
````

````python
quarterly_closing_aapl_2020[1, 2]
````

### Slicing

````python
# Get Closing Prices < 105
monthly_closings_aapl_2020[monthly_closings_aapl_2020 < 105]
````

````python
# Get Closing Prices > 120 & < 130
monthly_closings_aapl_2020[(monthly_closings_aapl_2020 > 120) & (monthly_closings_aapl_2020 < 130)]
````
