### Exercise 1: Stock Portfolio Analysis
#### Background
You are a portfolio analyst managing a portfolio of 5 stocks over 10 trading days. You need to analyze the daily prices and calculate various performance metrics.

#### Task 1.1: Create the Portfolio Data

Create a 2D NumPy array called `portfolio_prices` with shape (10, 5) representing:
- **Rows**: 10 trading days
- **Columns**: 5 different stocks (Stock A, B, C, D, E)

Use the following initial prices for Day 0:
- Stock A: 100, Stock B: 50, Stock C: 75, Stock D: 120, Stock E: 30

Generate random daily returns between -3% and +3% for each subsequent day using `np.random.uniform()`.

**Hint**: Use cumulative products to simulate realistic price movements.
```python

````python
np.random.seed(42)

initial_prices = np.array([100, 50, 75, 120, 30])

daily_returns = np.random.uniform(0.97, 1.03, size=(9, 5))

all_returns = np.vstack([np.ones((1, 5)), daily_returns])

portfolio_prices = initial_prices * all_returns

print(portfolio_prices.shape)
print(portfolio_prices)
````

#### Task 1.3: Calculate Daily Returns

Write a function `calculate_returns(prices)` that:
- Takes a 2D array of prices (shape: days × stocks)
- **Validates** that the input is a 2D NumPy array with at least 2 rows
- Calculates daily percentage returns for each stock
- Returns a 2D array with shape (days-1, stocks)

**Formula**: Return = (Price_today - Price_yesterday) / Price_yesterday * 100
```python

````python
portfolio_prices.shape
````

````python
import numpy as np

def calculate_returns(prices: np.ndarray) -> np.ndarray:
    
    if prices.ndim != 2:
        raise ValueError(f"Input must be 2D array, got {prices.ndim}D array")
    
    if prices.shape[0] < 2:
        raise ValueError(f"Input must have at least 2 rows, got {prices.shape[0]} row(s)")
    
    price_yesterday = prices[:-1, :]  # Alle Tage außer dem letzten
    price_today = prices[1:, :]       # Alle Tage außer dem ersten
    
    daily_returns = (price_today - price_yesterday) / price_yesterday * 100
    
    return daily_returns

returns = calculate_returns(portfolio_prices)
print(f"Shape of returns: {returns.shape}")
print(f"\nDaily returns (first 5 days):\n{returns[:5]}")

# Test 1: Nicht-NumPy array
# calculate_returns([[100, 50], [102, 51]])


# # # Test 2: 1D array
# calculate_returns(np.array([100, 102, 104]))

# # Test 3: Nur 1 Zeile
# calculate_returns(np.array([[100, 50, 75]]))
````
