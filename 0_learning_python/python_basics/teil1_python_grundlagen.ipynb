### 4.2 Strings zusammensetzen: `+` vs. f-Strings

Mit `+` kann man Strings verketten – das ist aber fehleranfällig. Besser sind **f-Strings**: ein `f` vor dem String, Variablen in `{}`.

````python
lecturer = "Kuruscu"
age = 40

# Verkettung mit + : fehleranfällig (fehlendes Leerzeichen, Zahl muss zu str werden)
print("Der Dozent" + " " + lecturer + "ist ... Jahre alt")

# f-String: übersichtlich
print(f"Der Dozent {lecturer} ist {age} Jahre alt")
````

### 4.3 Zahlen formatieren

| Format | Wirkung | Beispiel | Ausgabe |
|---|---|---|---|
| `:.2f` | 2 Nachkommastellen | `f"{120000.33123:.2f}"` | `120000.33` |
| `:,` | Tausendertrennzeichen | `f"{1000:,}"` | `1,000` |
| `:,.2f` | beides | `f"{152500:,.2f}"` | `152,500.00` |
| `:.1%` | als Prozent | `f"{0.07:.1%}"` | `7.0%` |
| `:+.2f` | mit Vorzeichen | `f"{3.5:+.2f}"` | `+3.50` |

Mit `"""..."""` schreibst du **mehrzeilige** Strings – ideal für Reports.

````python
ticker = "AAPL"
shares = 1000
price = 152.50
value = shares * price

report = f"""
Portfolio Position:
Ticker:      {ticker}
Shares:      {shares:,}
Price:       ${price:.2f}
Total Value: ${value:,.2f}
"""
print(report)
````

## 5. Vergleichs- & logische Operatoren

| Operator | Bedeutung | Beispiel |
|----------|-----------|----------|
| `==` | Gleich | `5 == 5` → `True` |
| `!=` | Ungleich | `5 != 3` → `True` |
| `>` | Größer als | `5 > 3` → `True` |

## 7. Bedingungen: `if` / `elif` / `else`

Mit Bedingungen trifft ein Programm **Entscheidungen**: Je nachdem, ob eine Bedingung `True` oder `False` ist, wird ein anderer Codeblock ausgeführt.

```python
if bedingung:
    # wird ausgeführt, wenn bedingung True ist
elif andere_bedingung:
    # wird geprüft, wenn die erste False war
else:
    # wird ausgeführt, wenn keine Bedingung zutraf
```
⚠️ Der Doppelpunkt `:` und die **Einrückung** (4 Leerzeichen) sind Pflicht.

````python
stock_price = 145.50
target_price = 150.00

if stock_price >= target_price:
    print("SELL signal: Target price reached")
else:
    print("HOLD signal: Target price not yet reached")
````

## 8. Schleifen: `for`, `enumerate`, `break`, `continue`

Mit einer `for`-Schleife führst du denselben Code für **jedes Element** einer Liste (oder eines Dictionaries) aus.

````python
daily_returns = [2.5, -1.2, 3.8]

for r in daily_returns:
    print(f"Rendite: {r:+.2f}%")
````

`enumerate()` liefert zusätzlich den Index – praktisch für eine Nummerierung:

````python
for i, r in enumerate(daily_returns, start=1):
    print(f"Tag {i}: {r:+.2f}%")
````

### Dictionaries durchlaufen
Ein Dictionary speichert **Schlüssel → Wert**-Paare. Mit `.items()` bekommst du beides:

````python
prices = {"AAPL": 178.50, "MSFT": 380.00, "NVDA": 485.20}

print(prices["AAPL"])   # Zugriff über den Schlüssel

for ticker, price in prices.items():
    print(f"{ticker}: ${price:.2f}")
````

### `continue` und `break`
- `continue` → überspringt den **Rest dieses Durchlaufs** und macht mit dem nächsten Element weiter
- `break` → beendet die **ganze Schleife** sofort

````python
prices = [150, 0, 162, -5, 171, 140]

for p in prices:
    if p <= 0:
        print("  ungültiger Kurs – übersprungen")
        continue
    if p > 170:
        print(f"  {p} über 170 – Schleife beendet")
        break
    print(f"Kurs: {p}")
````
