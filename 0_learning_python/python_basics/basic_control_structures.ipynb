# 1. CONDITIONAL STATEMENTS

### 1.1 Introduction to Conditionals
**What are conditionals?**
- Allow programs to make decisions based on conditions
- Execute different code blocks depending on whether conditions are True or False
- Essential for risk assessment, trading strategies, and portfolio management

````python
price = 99

if price < 150:
    print("Günstig")
else:
    print("Teuer")
````

````python
stock_price = 145.50
target_price = 150.00

if stock_price >= target_price:
    print("SELL signal: Target price reached")
else:
    print("HOLD signal: Target price not yet reached")
````

````python
credit_score = 720

if credit_score >= 750:
    rating = "Excellent"
    interest_rate = 3.5
elif credit_score >= 700:
    rating = "Good"
    interest_rate = 4.2
elif credit_score >= 650:
    rating = "Fair"
    interest_rate = 5.8

print(f"Credit Rating: {rating}")
print(f"Interest Rate: {interest_rate}%")
````

````python
price = 125.30
pe_ratio = 18.5
market_cap = 500_000_000_000  # $500B

# Multiple criteria for investment
if price < 150 and pe_ratio < 20 and market_cap > 100_000_000_000:
    print("✓ Stock meets ALL investment criteria")
    print(f"  Price: ${price}")
    print(f"  P/E Ratio: {pe_ratio}")
    print(f"  Market Cap: ${market_cap:,.0f}")
else:
    print("✗ Stock does not meet criteria")
````

# Exercise 3: Portfolio Risk Monitor with Stop-Loss (20 Min)

## Task Description

You are managing a portfolio and need to monitor each position for risk management. Your system should evaluate positions, trigger stop-loss orders when needed, skip invalid data, and create a report of actionable positions.

---

## Given Data
```python
portfolio = {
    "AAPL": {"current_price": 178.50, "purchase_price": 185.00, "shares": 50},
    "MSFT": {"current_price": 380.00, "purchase_price": 350.00, "shares": 30},
    "GOOGL": {"current_price": 142.30, "purchase_price": 155.00, "shares": 25},
    "TSLA": {"current_price": 242.80, "purchase_price": 245.00, "shares": 40},
    "AMZN": {"current_price": 0, "purchase_price": 145.00, "shares": 35},  # Invalid data
    "NVDA": {"current_price": 485.20, "purchase_price": 420.00, "shares": 20},
    "META": {"current_price": 325.50, "purchase_price": 340.00, "shares": 15},
    "NFLX": {"current_price": 445.00, "purchase_price": 480.00, "shares": 10},
    "AMD": {"current_price": -5.50, "purchase_price": 125.00, "shares": 45},  # Invalid data
    "JPM": {"current_price": 148.90, "purchase_price": 155.00, "shares": 60}
}
```

---

## Requirements

1. Create a **for loop** that iterates through the portfolio dictionary
2. For each position, implement the following logic using **if-elif-else statements**:
   - **Skip** positions with invalid data (current_price <= 0)
   - **Stop-Loss Alert**: If any position has a loss > 10%, trigger stop-loss and stop checking remaining positions
   - **Warning**: If loss is between 5% and 10% (inclusive)
   - **Hold**: If gain/loss is between -5% and +15%
   - **Take Profit**: If gain is > 15%
3. Calculate for each position:
   - Percentage change: `((current_price - purchase_price) / purchase_price) * 100`
   - Position value: `current_price * shares`
   - Profit/Loss in USD: `(current_price - purchase_price) * shares`
4. Print details for each processed position
---

## Expected Output Format
```
Processing AAPL...
  Current: $178.50 | Purchase: $185.00 | Change: -3.51%
  Position Value: $8,925.00 | P/L: -$325.00
  Action: HOLD

Processing MSFT...
  Current: $380.00 | Purchase: $350.00 | Change: +8.57%
  Position Value: $11,400.00 | P/L: +$900.00
  Action: HOLD
```

````python
portfolio = {
    "AAPL": {"current_price": 178.50, "purchase_price": 185.00, "shares": 50},
    "MSFT": {"current_price": 380.00, "purchase_price": 350.00, "shares": 30},
    "GOOGL": {"current_price": 142.30, "purchase_price": 155.00, "shares": 25},
    "TSLA": {"current_price": 242.80, "purchase_price": 245.00, "shares": 40},
    "AMZN": {"current_price": 0, "purchase_price": 145.00, "shares": 35},  # Invalid data
    "NVDA": {"current_price": 485.20, "purchase_price": 420.00, "shares": 20},
    "META": {"current_price": 325.50, "purchase_price": 340.00, "shares": 15},
    "NFLX": {"current_price": 445.00, "purchase_price": 480.00, "shares": 10},
    "AMD": {"current_price": -5.50, "purchase_price": 125.00, "shares": 45},  # Invalid data
    "JPM": {"current_price": 148.90, "purchase_price": 155.00, "shares": 60}
}

for ticker, data in portfolio.items():
    if data["current_price"] <= 0:
        continue
    pct_change = ((data["current_price"] - data["purchase_price"]) / data["purchase_price"]) * 100
    
    if pct_change < -10:
        break
    elif -10 <= pct_change < -5:
        action = "warning"
    elif -5 <= pct_change <= 15:
        action = "hold"
    elif pct_change > 15:
        action = "take"

    print(
    f"""
      Processing {ticker}...
      Current: ${data["current_price"]} | Purchase: ${data["purchase_price"]} | Change: {round(pct_change, 2)}%
      Position Value: $11,400.00 | P/L: +$900.00
      Action: {action.upper()}
    """
    )
````
