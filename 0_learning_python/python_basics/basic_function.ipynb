# Exercise 2: Portfolio Analysis Functions

## Objective
Build a modular portfolio analysis system using functions that call other functions. You will work with a dictionary containing portfolio data and create functions to analyze and format the results.

## Provided Data Structure

We provide you with the following portfolio dictionary:
```python
portfolio = {
    'AAPL': {'shares': 50, 'purchase_price': 150.00, 'current_price': 178.50},
    'MSFT': {'shares': 30, 'purchase_price': 280.00, 'current_price': 310.25},
    'GOOGL': {'shares': 20, 'purchase_price': 2800.00, 'current_price': 2650.75},
    'TSLA': {'shares': 15, 'purchase_price': 700.00, 'current_price': 245.30}
}
```

Each stock entry contains:
- `shares`: Number of shares owned
- `purchase_price`: Price per share when purchased
- `current_price`: Current market price per share

## Tasks

### Task 1: Basic Calculation Functions

Implement the following functions:

**a) `calculate_position_value(shares, current_price)`**
- Parameters: number of shares (int/float), current price per share (float)
- Returns: total current value of the position (float)

**b) `calculate_position_cost(shares, purchase_price)`**
- Parameters: number of shares (int/float), purchase price per share (float)
- Returns: total cost basis of the position (float)

**c) `calculate_position_gain_loss(shares, purchase_price, current_price)`**
- Parameters: number of shares, purchase price, current price
- Returns: absolute gain/loss in dollars (float)
- Hint: This function should call the two functions above

**d) `calculate_position_return(shares, purchase_price, current_price)`**
- Parameters: number of shares, purchase price, current price
- Returns: percentage return (float)
- Hint: Use the gain/loss function and cost function
