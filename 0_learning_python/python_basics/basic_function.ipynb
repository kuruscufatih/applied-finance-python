## Task 1: Calculate Compound Annual Growth Rate (CAGR)

Given a list of dictionaries containing stock data, calculate the CAGR for each stock using a loop.

Formula: CAGR = (final_value / initial_value)^(1/years) - 1

````python
a = dict(ticker = "AAPL")
a["ticker"]
````

````python
portfolio = [
    {"ticker": "AAPL", "initial_price": 100, "final_price": 180, "years": 5},
    {"ticker": "MSFT", "initial_price": 50, "final_price": 120, "years": 4},
    {"ticker": "GOOGL", "initial_price": 800, "final_price": 1500, "years": 6},
    {"ticker": "AMZN", "initial_price": 300, "final_price": 450, "years": 3},
    {"ticker": "TSLA", "initial_price": 200, "final_price": 250, "years": 2}
]

# TODO: WRITE YOUR FUNCTION
def calculate_cagr(final_price, initial_price, years):
    cagr = (final_price / initial_price) ** (1 / years) -1
    return cagr

# TODO: WRITE YOUR LOOP AND PRINT RESULTS FOR EACH TICKER
for ticker in portfolio:
    print(ticker["ticker"],
        calculate_cagr(ticker["final_price"], ticker["initial_price"], ticker["years"])
    )
````

## Task 2: Filter and Transform (10 min)

Using the same portfolio data, create a list comprehension that:
1. Only includes stocks with CAGR > 10%
2. Returns dictionaries with ticker, cagr, and rating where rating is:
   - "Excellent" if CAGR > 20%
   - "Good" if CAGR > 10%
## Expected Output
Top Performing Stocks:
AAPL: 12.47% - Good
MSFT: 24.57% - Excellent
GOOGL: 11.05% - Good
AMZN: 14.47% - Good
TSLA: 11.80% - Good
```

````python
# TODO: WRITE YOUR FUNCTION
def calculate_cagr(final_price, initial_price, years):
    cagr = (final_price / initial_price) ** (1 / years) -1
    return cagr

# TODO: WRITE YOUR LOOP AND PRINT RESULTS FOR EACH TICKER
for ticker in portfolio:
    cagr = calculate_cagr(ticker["final_price"], ticker["initial_price"], ticker["years"]) * 100
    cagr = round(cagr, 2)
    if cagr > 0.2:
        print(ticker["ticker"], cagr, "%", "Excellent")
    elif cagr > 0.1:
        print(ticker["ticker"], cagr, "%", "Good")
````
