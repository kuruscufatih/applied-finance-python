## Task 1: Calculate Compound Annual Growth Rate (CAGR)

Given a list of dictionaries containing stock data, calculate the CAGR for each stock using a list comprehension.

Formula: CAGR = (final_value / initial_value)^(1/years) - 1

````python
portfolio = [
    {"ticker": "AAPL", "initial_price": 100, "final_price": 180, "years": 5},
    {"ticker": "MSFT", "initial_price": 50, "final_price": 120, "years": 4},
    {"ticker": "GOOGL", "initial_price": 800, "final_price": 1500, "years": 6},
    {"ticker": "AMZN", "initial_price": 300, "final_price": 450, "years": 3},
    {"ticker": "TSLA", "initial_price": 200, "final_price": 250, "years": 2}
]

def calc_cagr(initial_price: int, final_price: int, years: int) -> float:

    return (final_price / initial_price) ** (1 / years) - 1


# TODO: Use list comprehension to create a list of dictionaries with ticker and cagr
cagr_results = [dict(ticker=stock["ticker"], cagr= calc_cagr(stock["initial_price"], stock["final_price"], stock["years"])) for stock in portfolio]

print("Stock CAGR Analysis:")
for stock in cagr_results:
    # print(f"{stock['ticker']}: {stock['cagr']:.2%}")
    print(stock)
````

## Task 2: Filter and Transform

Using the same portfolio data, create a list comprehension that:
1. Only includes stocks with CAGR > 10%
2. Returns dictionaries with ticker, cagr, and rating where rating is:
   - "Excellent" if CAGR > 20%
   - "Good" if CAGR > 10%
## Expected Output
```

AMZN: 14.47%
```

````python
# cagr_results 

def get_rating(cagr):
    """
    Assign a rating based on CAGR
    
    Parameters:
    - cagr: Compound Annual Growth Rate as decimal
    
    Returns:
    - Rating string
    """
    if cagr > 0.2:
        return "Excellent"
    elif cagr > 0.1:
        return "Good"
    else:
        return "Poor"
    
# TODO: Create list comprehension with filter and rating
top_performers = [{"ticker":p["ticker"],"cagr":p["cagr"],"rating":get_rating(p["cagr"])} for p in cagr_results if p["cagr"] > 0.10]
print("\nTop Performing Stocks:")
for stock in top_performers:
    print(f"{stock['ticker']}: {stock['cagr']:.2%} - {stock['rating']}")
````

### Task 1: Basic Calculation Functions

Implement the following functions:

**a) `calculate_position_value(shares, current_price)`**
- Parameters: number of shares (int/float), current price per share (float)
- Returns: total current value of the position (float)

**b) `calculate_position_cost(shares, purchase_price)`**
- Parameters: number of shares (int/float), purchase price per share (float)
- Returns: total cost basis of the position (float)

**c) `calculate_position_gain_loss(shares, purchase_price, current_price)`**
- Parameters: number of shares, purchase price, current price
- Returns: absolute gain/loss in dollars (float)
- Hint: This function should call the two functions above

**d) `calculate_position_return(shares, purchase_price, current_price)`**
- Parameters: number of shares, purchase price, current price
- Returns: percentage return (float)
- Hint: Use the gain/loss function and cost function

````python
def calculate_position_value(shares: float, current_price: float) -> float:
    return shares * current_price

def calculate_position_cost(shares: float, purchase_price: float) -> float:
    return shares * purchase_price

def calculate_position_gain_loss(shares: float, purchase_price: float, current_price: float) -> float:
    return calculate_position_value(shares, current_price) - calculate_position_cost(shares, purchase_price)

def calculate_position_return(shares: float, purchase_price: float, current_price: float) -> float:
    gain_loss = calculate_position_gain_loss(shares, purchase_price, current_price)
    cost = calculate_position_cost(shares, purchase_price)
    return (gain_loss / cost) * 100 if cost != 0 else 0

calculate_position_return(50, 150, 220)
````

### Task 2: Portfolio Analysis Function

**` def analyze_portfolio(portfolio)`**
- Parameter: portfolio dictionary (as provided above)
- Returns: dictionary with the following structure:
```python
{
    'total_invested': float,      # Total amount invested across all positions
    'total_current_value': float, # Current total value of portfolio
    'total_gain_loss': float,     # Total gain/loss in dollars
    'total_return': float,        # Overall portfolio return in %
    'positions': {
        'AAPL': {
            'current_value': float,
            'gain_loss': float,
            'return_pct': float
        },
        # ... same for other stocks
    }
}
```

This function must call your calculation functions from Task 1.

````python
portfolio = {
    'AAPL': {'shares': 50, 'purchase_price': 150.00, 'current_price': 178.50},
    'MSFT': {'shares': 30, 'purchase_price': 280.00, 'current_price': 310.25},
    'GOOGL': {'shares': 20, 'purchase_price': 2800.00, 'current_price': 2650.75},
    'TSLA': {'shares': 15, 'purchase_price': 700.00, 'current_price': 245.30}
}
````

````python
def analyze_portfolio(portfolio: dict) -> dict:
    total_invested = 0
    total_current_value = 0

    for ticker, data in portfolio.items():
        total_invested += calculate_position_cost(data["shares"], data["purchase_price"])
        total_current_value += calculate_position_value(data["shares"], data["current_price"])

    total_gain_loss = total_current_value - total_invested
    total_return = (total_gain_loss / total_invested) * 100

    return {
        'total_invested': total_invested,      # Total amount invested across all positions
        'total_current_value': total_current_value, # Current total value of portfolio
        'total_gain_loss': total_gain_loss,     # Total gain/loss in dollars
        'total_return': total_return, 
        'positions': {
            key:{'current_value': calculate_position_value(portfolio[key]['shares'], portfolio[key]['current_price']), 'gain_loss': 0, 'return_pct': 0} for key in portfolio.keys()}
    }
                                  
````
