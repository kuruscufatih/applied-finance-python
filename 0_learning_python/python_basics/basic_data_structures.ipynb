```python
# Dictionaries in Python - Applied Finance
# Course: Wirtschaftsinformatik - Applied Finance in Python

"""
Learning Objectives:
1. Understand dictionary data structures
2. Create and manipulate dictionaries
3. Apply dictionaries to financial data
4. Work with nested dictionaries for complex financial structures
"""

# =============================================================================
# 1. INTRODUCTION TO DICTIONARIES
# =============================================================================

print("="*60)
print("1. WHAT ARE DICTIONARIES?")
print("="*60)

# Dictionaries store key-value pairs
# Syntax: {key1: value1, key2: value2, ...}

# Example: Stock information
stock = {
    "ticker": "AAPL",
    "price": 178.50,
    "currency": "USD",
    "exchange": "NASDAQ"
}

print(f"Stock dictionary: {stock}")
print(f"Type: {type(stock)}")

# =============================================================================
# 2. CREATING DICTIONARIES
# =============================================================================

print("\n" + "="*60)
print("2. CREATING DICTIONARIES")
print("="*60)

# Method 1: Literal notation
portfolio = {
    "AAPL": 10,
    "MSFT": 15,
    "GOOGL": 5
}

# Method 2: dict() constructor
bond = dict(issuer="US Treasury", coupon=4.5, maturity=2030)

# Method 3: Empty dictionary
transactions = {}

print(f"Portfolio: {portfolio}")
print(f"Bond: {bond}")
print(f"Empty transactions: {transactions}")

# =============================================================================
# 3. ACCESSING VALUES
# =============================================================================

print("\n" + "="*60)
print("3. ACCESSING DICTIONARY VALUES")
print("="*60)

stock_data = {
    "ticker": "TSLA",
    "price": 242.80,
    "volume": 125000000,
    "market_cap": 770e9
}

# Access using keys
print(f"Ticker: {stock_data['ticker']}")
print(f"Price: ${stock_data['price']}")

# Safe access with .get() method
print(f"Volume: {stock_data.get('volume')}")
print(f"Dividend: {stock_data.get('dividend', 'N/A')}")  # Default value

# =============================================================================
# 4. MODIFYING DICTIONARIES
# =============================================================================

print("\n" + "="*60)
print("4. MODIFYING DICTIONARIES")
print("="*60)

portfolio = {"AAPL": 10, "MSFT": 15}
print(f"Original portfolio: {portfolio}")

# Adding new key-value pairs
portfolio["GOOGL"] = 5
print(f"After adding GOOGL: {portfolio}")

# Updating existing values
portfolio["AAPL"] = 12
print(f"After updating AAPL: {portfolio}")

# Removing items
del portfolio["MSFT"]
print(f"After deleting MSFT: {portfolio}")

# Remove and return value
googl_shares = portfolio.pop("GOOGL")
print(f"Popped GOOGL: {googl_shares} shares")
print(f"Portfolio now: {portfolio}")

# =============================================================================
# 5. DICTIONARY METHODS
# =============================================================================

print("\n" + "="*60)
print("5. USEFUL DICTIONARY METHODS")
print("="*60)

prices = {
    "AAPL": 178.50,
    "MSFT": 378.90,
    "GOOGL": 141.80,
    "AMZN": 178.25
}

# Get all keys
print(f"Tickers: {list(prices.keys())}")

# Get all values
print(f"Prices: {list(prices.values())}")

# Get all items (key-value pairs)
print(f"Items: {list(prices.items())}")

# Check if key exists
print(f"Is AAPL in portfolio? {'AAPL' in prices}")
print(f"Is TSLA in portfolio? {'TSLA' in prices}")

# =============================================================================
# 6. ITERATING THROUGH DICTIONARIES
# =============================================================================

print("\n" + "="*60)
print("6. ITERATING THROUGH DICTIONARIES")
print("="*60)

portfolio_value = {
    "AAPL": 1785.00,
    "MSFT": 5683.50,
    "GOOGL": 709.00
}

# Iterate through keys
print("\nPortfolio holdings:")
for ticker in portfolio_value:
    print(f"  {ticker}")

# Iterate through values
print("\nPortfolio values:")
for value in portfolio_value.values():
    print(f"  ${value:,.2f}")

# Iterate through key-value pairs
print("\nComplete portfolio:")
total = 0
for ticker, value in portfolio_value.items():
    print(f"  {ticker}: ${value:,.2f}")
    total += value

print(f"\nTotal portfolio value: ${total:,.2f}")

# =============================================================================
# 7. NESTED DICTIONARIES - FINANCIAL DATA STRUCTURES
# =============================================================================

print("\n" + "="*60)
print("7. NESTED DICTIONARIES")
print("="*60)

# Complex financial data structure
portfolio_detailed = {
    "AAPL": {
        "shares": 10,
        "purchase_price": 150.00,
        "current_price": 178.50,
        "sector": "Technology"
    },
    "JPM": {
        "shares": 20,
        "purchase_price": 145.00,
        "current_price": 158.30,
        "sector": "Finance"
    }
}

# Access nested data
print(f"AAPL shares: {portfolio_detailed['AAPL']['shares']}")
print(f"JPM sector: {portfolio_detailed['JPM']['sector']}")

# Calculate gains
for ticker, data in portfolio_detailed.items():
    shares = data['shares']
    purchase = data['purchase_price']
    current = data['current_price']
    gain = (current - purchase) * shares
    gain_pct = ((current - purchase) / purchase) * 100
    
    print(f"\n{ticker}:")
    print(f"  Gain: ${gain:.2f} ({gain_pct:.2f}%)")

# =============================================================================
# 8. PRACTICAL EXAMPLE: PORTFOLIO ANALYSIS
# =============================================================================

print("\n" + "="*60)
print("8. PRACTICAL EXAMPLE: PORTFOLIO ANALYSIS")
print("="*60)

# Portfolio with multiple stocks
portfolio = {
    "AAPL": {"shares": 10, "price": 178.50},
    "MSFT": {"shares": 15, "price": 378.90},
    "GOOGL": {"shares": 5, "price": 141.80},
    "AMZN": {"shares": 8, "price": 178.25}
}

# Calculate portfolio metrics
total_value = 0
positions = []

print("\nPortfolio Summary:")
print("-" * 50)

for ticker, data in portfolio.items():
    shares = data["shares"]
    price = data["price"]
    position_value = shares * price
    total_value += position_value
    
    positions.append({
        "ticker": ticker,
        "value": position_value
    })
    
    print(f"{ticker:6s} | {shares:3d} shares @ ${price:7.2f} = ${position_value:10,.2f}")

print("-" * 50)
print(f"{'TOTAL':6s} | {' '*20} ${total_value:10,.2f}")

# Calculate portfolio weights
print("\nPortfolio Weights:")
print("-" * 50)
for position in positions:
    weight = (position["value"] / total_value) * 100
    print(f"{position['ticker']:6s} | {weight:5.2f}%")

# =============================================================================
# 9. DICTIONARY COMPREHENSIONS
# =============================================================================

print("\n" + "="*60)
print("9. DICTIONARY COMPREHENSIONS")
print("="*60)

# Create dictionary from lists
tickers = ["AAPL", "MSFT", "GOOGL"]
prices = [178.50, 378.90, 141.80]

price_dict = {ticker: price for ticker, price in zip(tickers, prices)}
print(f"Price dictionary: {price_dict}")

# Transform dictionary values
price_eur = {ticker: price * 0.92 for ticker, price in price_dict.items()}
print(f"Prices in EUR: {price_eur}")

# Filter dictionary
expensive_stocks = {t: p for t, p in price_dict.items() if p > 200}
print(f"Stocks over $200: {expensive_stocks}")

# =============================================================================
# 10. EXERCISES
# =============================================================================

print("\n" + "="*60)
print("10. EXERCISES FOR STUDENTS")
print("="*60)

print("""
Exercise 1: Create a Bond Portfolio
- Create a dictionary for 3 different bonds
- Each bond should have: issuer, coupon_rate, maturity_year, face_value
- Calculate the annual coupon payment for each bond

Exercise 2: Currency Converter
- Create a dictionary with exchange rates (USD to EUR, GBP, JPY, CHF)
- Write a function that converts an amount from USD to any currency
- Test with converting $1000 to each currency

Exercise 3: Stock Performance Tracker
- Create a nested dictionary with 5 stocks
- Each stock should have: purchase_price, current_price, quantity
- Calculate:
  a) Total investment (initial)
  b) Current portfolio value
  c) Total gain/loss
  d) Best performing stock (highest % gain)
  
Exercise 4: Transaction History
- Create a list of transaction dictionaries
- Each transaction: date, ticker, type (buy/sell), shares, price
- Calculate the current position for each ticker

Exercise 5: Financial Ratios
- Create a dictionary with company financial data:
  revenue, net_income, total_assets, total_equity, current_assets, current_liabilities
- Calculate and store in a new dictionary:
  - Profit Margin
  - ROE (Return on Equity)
  - ROA (Return on Assets)
  - Current Ratio
""")

# =============================================================================
# SOLUTIONS (for instructor)
# =============================================================================

# Exercise 1 Solution
print("\nExercise 1 - Sample Solution:")
bonds = {
    "US_10Y": {"issuer": "US Treasury", "coupon": 4.5, "maturity": 2034, "face_value": 1000},
    "DE_10Y": {"issuer": "German Bund", "coupon": 2.8, "maturity": 2034, "face_value": 1000},
    "CORP_A": {"issuer": "Apple Inc", "coupon": 3.2, "maturity": 2030, "face_value": 1000}
}

for bond_id, bond in bonds.items():
    annual_payment = bond["face_value"] * (bond["coupon"] / 100)
    print(f"{bond_id}: Annual coupon = ${annual_payment:.2f}")

print("\n" + "="*60)
print("END OF NOTEBOOK")
print("="*60)
```
