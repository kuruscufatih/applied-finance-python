#### Accessing Values

```python
portfolio = {
    'AAPL': 178.50,
    'MSFT': 338.11,
    'GOOGL': 125.30
}
```

##### Method 1: Square brackets

```python
apple_price = portfolio['MSFT'] #Throws error if not available
print(f"\nApple price: ${apple_price}")
```

```python
apple_price = portfolio['TSLA'] #Throws error if not available
print(f"\nApple price: ${apple_price}")
```

##### Method 2: .get() method (safer!)

```python
tesla_price = portfolio.get('TSLA')
print(f"Tesla price (not in portfolio): {tesla_price}")

# .get() with default value
tesla_price_default = portfolio.get('TSLA', 0.0)
print(f"Tesla price with default: ${tesla_price_default}")
```

#### Modifying Dictionaries

```python
# portfolio = {
#     'AAPL': 178.50,
#     'MSFT': 338.11,
#     'GOOGL': 125.30
# }
```

```python
portfolio = {'AAPL': 178.50, 'MSFT': 338.11}
portfolio['GOOGL'] = 125.30
print("\nAfter adding Google:", portfolio)
```

```python
portfolio['AAPL'] = 180.00
print("After updating Apple:", portfolio)
```

```python
# Removing entries
del portfolio['MSFT']
print("After deleting Microsoft:", portfolio)
```

```python
# Using .pop() (returns the value)
removed_price = portfolio.pop('GOOGL')
print(f"Removed Google at ${removed_price}")
print("Final portfolio:", portfolio)
```
