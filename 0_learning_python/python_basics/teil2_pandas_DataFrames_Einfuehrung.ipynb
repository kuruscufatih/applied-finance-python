## 1 · Excel und pandas im Vergleich

Vieles, was ihr aus Excel kennt, gibt es in pandas auch, nur eben als Code:

| In Excel | In pandas |
|---|---|
| Tabellenblatt | `DataFrame` |
| Eine Spalte | `Series` |
| Zeilennummern (1, 2, 3 …) | `index` |
| Spaltenüberschriften | `columns` |
| Formel eintragen und nach unten ziehen | `df["Neu"] = df["A"] * df["B"]` |
| Filter setzen | `df[df["Kurs"] > 100]` |
| Sortieren | `df.sort_values("Kurs")` |
| Pivot-Tabelle | `df.groupby("Branche").sum()` |
| CSV-Datei öffnen | `pd.read_csv("datei.csv")` |

## 2 · Einen DataFrame erstellen

Der einfachste Weg: ein **Dictionary**, bei dem jeder Schlüssel ein Spaltenname ist und jeder Wert eine Liste mit den Einträgen dieser Spalte.

Als Beispiel nehmen wir ein kleines Aktiendepot.

⚠️ Die Zahlen in diesem Notebook sind ausgedacht und dienen nur zur Übung.

```python
portfolio = pd.DataFrame({
    "Company":   ["Allianz", "SAP", "Infineon", "Deutsche Telekom", "Munich Re"],
    "Ticker":    ["ALV.DE", "SAP.DE", "IFX.DE", "DTE.DE", "MUV2.DE"],
    "Sector":    ["Insurance", "Technology", "Technology", "Telecom", "Insurance"],
    "Shares":    [10, 15, 50, 100, 4],
    "BuyPrice":  [195.00, 110.00, 30.00, 19.00, 370.00],
    "Price":     [413.00, 186.00, 38.50, 27.00, 520.00],
})

portfolio
```

💡 Links seht ihr den **Index** (0, 1, 2 …). Er nummeriert die Zeilen, ähnlich wie die Zeilennummern in Excel.

## 4 · Spalten, Zeilen und einzelne Werte auswählen

### Eine Spalte

Eine Spalte wählt ihr mit eckigen Klammern und dem Spaltennamen aus. Das Ergebnis ist eine **Series**, also eine einzelne Spalte mit Index.

```python
portfolio["Price"]
```

### Mehrere Spalten

Für mehrere Spalten übergebt ihr eine **Liste** von Namen, deshalb die doppelten Klammern.

```python
portfolio[["Company", "Price"]]
```

### Zeilen mit `iloc` und `loc`

- `iloc` wählt nach **Position** aus: die erste Zeile ist 0, die letzte ist -1.
- `loc` wählt nach **Index-Bezeichnung** aus. Das wird besonders bei Datumswerten wichtig (Teil 9).

```python
portfolio.iloc[0]        # first row
```

```python
portfolio.iloc[-1]       # last row
```

### Ein einzelner Wert

```python
portfolio["Price"].iloc[0]     # price of the first stock
```

## 5 · Filtern

Filtern funktioniert in zwei Schritten. Zuerst stellt ihr eine Bedingung, das ergibt für jede Zeile `True` oder `False`:

```python
portfolio["Price"] > 100
```

Diese Liste aus `True` und `False` setzt ihr dann in eckige Klammern. Übrig bleiben nur die Zeilen mit `True`:

```python
portfolio[portfolio["Price"] > 100]
```

```python
portfolio[portfolio["Sector"] == "Technology"]
```

Mehrere Bedingungen verbindet ihr mit `&` (und) oder `|` (oder). Jede Bedingung muss dabei in runden Klammern stehen.

```python
portfolio[(portfolio["Sector"] == "Insurance") & (portfolio["Price"] > 450)]
```

## 6 · Neue Spalten berechnen

Das ist die eigentliche Stärke von pandas. Eine Rechnung mit Spalten wird **automatisch für jede Zeile** ausgeführt. In Excel würdet ihr eine Formel schreiben und nach unten ziehen, hier reicht eine Zeile.

```python
portfolio["Value"] = portfolio["Shares"] * portfolio["Price"]
portfolio["Invested"] = portfolio["Shares"] * portfolio["BuyPrice"]
portfolio["Return"] = portfolio["Price"] / portfolio["BuyPrice"] - 1

portfolio
```

💡 Mit `round` und `style.format` wird die Ausgabe lesbarer. Die Daten selbst bleiben dabei unverändert, es ändert sich nur die Anzeige.

```python
portfolio.style.format({
    "BuyPrice": "{:.2f} €", "Price": "{:.2f} €",
    "Value": "{:,.0f} €", "Invested": "{:,.0f} €",
    "Return": "{:.1%}",
})
```

✏️ **Probiert es selbst 3:** Berechnet eine neue Spalte `Profit` mit dem Gewinn in Euro (aktueller Wert minus investierter Betrag).

## 7 · Sortieren und Ranglisten

```python
portfolio.sort_values("Return", ascending=False)    # highest return first
```

```python
portfolio.nlargest(3, "Value")                      # the 3 largest positions
```

```python
portfolio["Rank"] = portfolio["Return"].rank(ascending=False).astype(int)
portfolio[["Company", "Return", "Rank"]]
```

## 8 · Zusammenfassen und Gruppieren

Einzelne Kennzahlen für eine Spalte:

```python
total_value = portfolio["Value"].sum()
total_invested = portfolio["Invested"].sum()

print(f"Portfolio value: {total_value:,.2f} €")
print(f"Invested:        {total_invested:,.2f} €")
print(f"Total return:    {total_value / total_invested - 1:.1%}")
```

Mit `groupby` fasst ihr nach einer Spalte zusammen, ähnlich wie bei einer Pivot-Tabelle in Excel:

```python
portfolio.groupby("Sector")[["Invested", "Value"]].sum()
```

💡 Daraus lässt sich direkt berechnen, wie viel Prozent des Depots auf jede Branche entfallen:

```python
by_sector = portfolio.groupby("Sector")["Value"].sum()
(by_sector / by_sector.sum()).sort_values(ascending=False)
```

```python
by_sector.plot(kind="bar", title="Portfolio value by sector (€)");
```

```python
dates = pd.bdate_range("2024-01-01", "2025-12-31")        # business days (Mon-Fri)
rng = np.random.default_rng(42)
daily_moves = rng.normal(0.0004, 0.015, len(dates))      # random daily changes
prices = pd.DataFrame(
    {"Close": 50 * np.cumprod(1 + daily_moves)},
    index=dates,
)
prices.index.name = "Date"

prices.head()
```

### Nach Datum auswählen

Mit `loc` könnt ihr direkt nach Datum oder Zeitraum auswählen. Ein unvollständiges Datum wie `"2025-03"` wählt den ganzen Monat aus.

```python
prices.loc["2025-03-14"]
```

```python
prices.loc["2025-03"]                     # whole month of March 2025
```

```python
prices.loc["2025-06-01":"2025-06-10"]     # a date range
```

### Erster und letzter Kurs, Rendite und CAGR

Damit habt ihr alles für die Rechnung aus Lernfall 1:

```python
first = prices["Close"].iloc[0]
last = prices["Close"].iloc[-1]
years = (prices.index[-1] - prices.index[0]).days / 365.25

print(f"Total return: {last / first - 1:.1%}")
print(f"CAGR:         {(last / first) ** (1 / years) - 1:.1%}")
```

### Tägliche Renditen mit `pct_change`

`pct_change()` berechnet für jede Zeile die Veränderung zum Vortag in Prozent. Der erste Wert ist leer (`NaN`), weil es keinen Vortag gibt.

```python
prices["Daily return"] = prices["Close"].pct_change()
prices.head()
```

```python
prices["Daily return"].describe()
```

### Von Tagen auf Monate: `resample`

`resample("ME")` fasst die Daten pro Monat zusammen. Mit `.last()` nehmt ihr den letzten Kurs jedes Monats.

```python
monthly = prices["Close"].resample("ME").last()
monthly.head()
```

### Gleitender Durchschnitt mit `rolling`

Ein gleitender Durchschnitt glättet den Kursverlauf. `rolling(50).mean()` berechnet für jeden Tag den Durchschnitt der letzten 50 Handelstage.

```python
prices["MA50"] = prices["Close"].rolling(50).mean()
prices[["Close", "MA50"]].plot(title="Beispiel AG: price and 50-day moving average");
```
