### 2.3. Operators - Comparison & Logical

## Comparison Operators
Comparison operators return **Boolean values** (`True` or `False`):

| Operator | Bedeutung | Beispiel |
|----------|-----------|----------|
| `==` | Gleich | `5 == 5` → `True` |
| `!=` | Ungleich | `5 != 3` → `True` |
| `>` | Größer als | `5 > 3` → `True` |
| `<` | Kleiner als | `3 < 5` → `True` |
| `>=` | Größer oder gleich | `5 >= 5` → `True` |
| `<=` | Kleiner oder gleich | `3 <= 5` → `True` |

## Logical Operators
Logische Operatoren verknüpfen Bedingungen:

- **`and`** - Beide Bedingungen müssen `True` sein
- **`or`** - Mindestens eine Bedingung muss `True` sein
- **`not`** - Negiert die Bedingung

```python
# Aktienkurs-Entscheidungslogik
stock_price = 152.50
buy_limit = 150
sell_limit = 160
is_market_open = True
emergency_exit = False


# Verkaufen, wenn Preis über Limit ODER Notfall
sell_limit < stock_price or emergency_exit
```
